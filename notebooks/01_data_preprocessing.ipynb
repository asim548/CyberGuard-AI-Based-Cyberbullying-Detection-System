# CyberGuard – Week 1: Data Preprocessing

This notebook loads the cyberbullying dataset, explores it, and preprocesses text using NLTK.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_dataset
from src.preprocess import preprocess_text
from config import LABELS
```

```python
df = load_dataset()
print(f"Samples: {len(df):,}")
df.head(10)
```

```python
df['label'].value_counts().plot(kind='bar', color=['#22c55e', '#f59e0b', '#ef4444'])
plt.title('Class Distribution (3 Categories)')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
df['text_length'] = df['text'].str.len()
sns.histplot(df['text_length'], bins=40, kde=True)
plt.title('Tweet / Comment Length Distribution')
plt.show()
```

```python
df['processed'] = df['text'].apply(preprocess_text)
df[['text', 'processed', 'label']].head(10)
```

```python
out_path = ROOT / 'data' / 'processed_sample.csv'
df[['text', 'processed', 'label']].to_csv(out_path, index=False)
print(f'Saved to {out_path}')
```
